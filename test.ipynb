- Happy number  19
- 1² + 9² = 82
- 8² + 2² = 68
- 6² + 8² = 100
- 1² + 0² + 0² = 1

```python
def is_happy(n):
    seen_number = set()
    while (n != 1) and (n not in seen_number):
        seen_number.add(n)
        n = sum([int(number) ** 2 for number in str(n)])
    
    return n == 1

is_happy(100)
```
