```python
def is_happy(n):
    seen_number = set()
    while (n != 1) and (n not in seen_number):
        n = sum([int(number) ** 2 for number in set(n)])
        seen_number.add(n)
    
    return n == 1

num = input("n? ")
is_happy(num)
```

```python
def number(n):
    total = 0
    while True:
        p = [int(number) for number in str(n)]
        for i in p:
            total += i ** 2  
            
        if total == 1 :
            break
        number(total)

        

number(19)
```
